```python
# flatten() vs reshape(-1)
```

```python
matrix = np.array([
    [1, 2],
    [3, 4]
])
```

```python
# Example 1 — flatten() (Independent Copy)
flat = matrix.flatten()
flat[0] = 100
```

```python
print(flat)
print('')
print(matrix)
```

```python
# Example 2 — reshape(-1) (View)

view = matrix.reshape(-1)
view[0] = 100
```

```python
print(view)
print('')
print(matrix) #conclusion reshape(-1) usually the view of Original Data
```
