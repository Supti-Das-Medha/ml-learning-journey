```python
#np.ones()
#np.ones(number_of_elements)
arr = np.ones(5)
print(arr)
```

```python
arr = np.ones(5,dtype=int) # Create a one-dimensional array filled with integer ones
print(arr) #print
print("ones: ",arr) #print with comment
```

```python
# np.arange()
#np.arange(start, stop, step)
# start : Starting value (included)
# stop  : Ending value (not included)
# step  : Interval between consecutive values

arr = np.arange(1,6) # Generate a sequence of integers from 1 to 5
print(arr) #print(Note: The stop value is not included in the output)
```

```python
arr = np.arange(0, 10, 2)
print(arr)
print("arange: ",arr)
```

```python
#np.linspace()
#np.linspace(start, stop, number_of_values)
# start            : Starting value (included)
# stop             : Ending value (included)
# number_of_values : Total number of values to generate

arr =  np.linspace(0, 10, 5) # Generate five evenly spaced values from 0 to 10
print(arr) #print
print("linspace: ",arr) #print with a descriptive label
```
